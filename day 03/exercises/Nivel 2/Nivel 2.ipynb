### Raspagem Avançada de Dados ✂️

##### BeautifulSoup

<pre>Em vez de capturarmos o padrão de formação do hyperlink para baixar os arquivos, vamos raspar a página por hyperlinks de arquivos <b>CSV</b>, diretamente da estrutura do <b>HTML</b>.</pre>

<pre>Portanto, usando a biblioteca <a href="https://beautiful-soup-4.readthedocs.io/en/latest/">BeautifulSoup</a>, crie uma lista (de nome <b>hyperlinks</b>) com todos os hyperlinks de arquivos <b>.csv</b> da páginas de <a href="https://www.gov.br/produtividade-e-comercio-exterior/pt-br/assuntos/comercio-exterior/estatisticas/base-de-dados-bruta">Estatísticas de Comércio Exterior</a>.</pre>

<details>
  <summary>Resposta</summary>

<br/>

```python
import requests
from bs4 import BeautifulSoup

url = "https://www.gov.br/produtividade-e-comercio-exterior/pt-br/assuntos/comercio-exterior/estatisticas/base-de-dados-bruta"

response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")

links = [link["href"] for link in soup.find_all("a", href=True) if ".csv" in link["href"]]
```

<br/>

</details>

<pre>A partir desta lista de hyperlinks, crie uma lista, chamada <b>arquivos</b>, com os nomes dos arquivos .csv, incluindo a extensão.</pre>

👉 dica: veja a documentação da função <a href="https://docs.python.org/3/library/os.path.html#os.path.basename">basename</a> do módulo `path` da biblioteca `os`

<details>
  <summary>Resposta</summary>

<br/>

```python
import os

arquivos = [os.path.basename(link) for link in links] 
```

##### Selenium

<pre>Vamos precisar baixar algumas tabelas de correlação para podermos avançar na exploração dos dados de comércio exterior brasileiros.</pre>

<pre>Na aula passada, nós baixamos e salvamos os arquivos acessando diretamente os hyperlinks com o método <b>get</b> da biblioteca <b>requests</b>.</pre>

<pre>Hoje, vamos praticar o uso da ferramenta Selenium. Vamos criar uma rotina para automatizar o download de algumas tabelas da página.</pre>

<pre>O código a seguir carrega os pacotes necessários, define a <b>url</b>, e inicia uma sessão no navegador Chrome. Em seguida, o Selenium recarrega a página para desativar a sobretela que direciona ao login, e clica no botão para aceitar os cookies, caso tenha aparecido.</pre>

````python
import os, time
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By

# cria as configurações do navegador para que o download seja realizado na pasta data do diretório corrente
options = Options()
folderpath = os.path.join(os.getcwd(), "data/")
if not os.path.exists('data'): os.makedirs('data')
prefs = {'download.default_directory' : folderpath}
options.add_experimental_option('prefs', prefs)

# inicia o navegador Chrome com as configurações acima
driver = webdriver.Chrome(options=options) 

# acessa o sítio de Dados Abertos das Estatísticas de Comércio Exterior.
url = 'https://www.gov.br/produtividade-e-comercio-exterior/pt-br/assuntos/comercio-exterior/estatisticas/base-de-dados-bruta'
driver.get(url)

# recarrega a página para desativar a sobretela que direciona ao login
time.sleep(1)
driver.refresh()

# clica no botão para aceitar os cookies, caso tenha aparecido
time.sleep(1)
if driver.find_element(By.CLASS_NAME, "btn-accept").is_displayed(): 
    driver.find_element(By.CLASS_NAME, "btn-accept").click()
````

<pre>Encontre os hyperlinks dos arquivos armazenados na lista <b>arquivos</b>, por meio dos <b>XPATH</b>, e instrua o driver do Chrome a clicar no hyperlink para realizar o download de cada arquivo.</pre>

👉 dica: lembre-se do <a href="https://devhints.io/xpath">cheat sheet</a> que vimos na aula de hoje.

👉 mais dica: para dar tempo de realizar o download de cada arquivo, aguarde um segundo entre downloads sucessivos.

<details>
  <summary>Resposta</summary>

<br/>

```python
for arquivo in arquivos:
    driver.find_element(By.XPATH, f"//a[contains(@href, '{arquivo}')]").click()
    time.sleep(1)
```
