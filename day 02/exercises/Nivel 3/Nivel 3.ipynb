## Sistema de Arquivos 🗂️

<pre>Um arquivo de texto pode ser uma fonte de dados ainda que não esteja em formato estruturado em linhas e colunas.</pre>

<pre>Vamos analisar as palavras mais frequentes de um arquivo de texto.</pre>

<pre>Mais à frente, usaremos a função abaixo para converter o texto para letras em caixa baixa e remover caracteres especiais, como pontuação. Portanto, execute a célula para ter a função à sua disposição.</pre>

````python
import re

def limpa_texto(linha):
    return ' '.join(re.sub(r'[^\w\s]', '', linha.lower() if type(linha) == str else '').split())
````

<pre>Para começar, construa uma função com o nome <b>calcula_frequencia</b>, que recebe uma lista de palavras e retorna um dicionário com a frequência de cada uma das palavras da lista.</pre>

<pre>Por exemplo:</pre>

```python
    calcula_frequencia(['um', 'dois', 'dois'])
``` 

<pre>retornará:</pre>

 ```python
     {'um': 1, 'dois': 2}
 ```

<details>
  <summary>Resposta</summary>

<br/>

```python
def calcula_frequencia(lista_itens=None):
    frequencia = dict()
    for item in lista_itens:
        frequencia[item] = frequencia.get(item, 0) + 1
    return frequencia
```

  <summary>Resposta</summary>

<br/>

```python
preco_de_gasolina_em_ubatuba = []

for linha in linhas:
    if linha[indice_municipio] == 'UBATUBA':
        if linha[indice_produto] == 'GASOLINA':
            preco = linha[indice_valor_de_venda].replace('.', '').replace(',', '.')
            preco = float(preco)
            preco_de_gasolina_em_ubatuba.append(preco)

preco_medio_de_gasolina_em_ubatuba = sum(preco_de_gasolina_em_ubatuba)/len(preco_de_gasolina_em_ubatuba)

f'Preço Médio de Gasolina em Ubatuba: R$ {preco_medio_de_gasolina_em_ubatuba: ,.2f}'
```

# ⚡️ Desafio: Raspagem de Dados ✂️

<pre>Considerando o padrão do formato da URL em que os dados da base de dados de <b>Combustíveis automotivos</b> da <a href="https://www.gov.br/anp/pt-br/centrais-de-conteudo/dados-abertos/serie-historica-de-precos-de-combustiveisl">"Série Histórica de Preços de Combustíveis e de GLP"</a> da Agência Nacional do Petróleo, Gás Natural e Biocombustíveis foram disponibilizados, baixe todos os arquivos referentes aos <b>combustíveis automotivos</b> dos anos de <b>2020</b> e <b>2021</b> utilizando estrutura de repetição <b>for</b> e interpolação de URL.</pre>

<details>
  <summary>Resposta</summary>

<br/>

```python
import requests

for ano in range(2020, 2022):
    for mes in range(1, 3):
        url = f"https://www.gov.br/anp/pt-br/centrais-de-conteudo/dados-abertos/arquivos/shpc/dsas/ca/ca-{ano}-{mes:02d}.csv"
        
        arquivo = requests.get(url)
        if arquivo.status_code == 200:
            
            conteudo = arquivo.content
            with open(f'ca-{ano}-{mes:02d}.csv', 'wb') as csv_file:
                csv_file.write(conteudo)
```
