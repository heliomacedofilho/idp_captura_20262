### Sistema de Arquivos 🗂️

<pre>A seguir, preparamos a função <b>download_file</b>, em que você passa a URL do aquivo <b>CSV</b> como parâmetro, e a função baixa o arquivo, com a mesma nomenclatura e extensão, para a pasta <b>data</b>, que figurará no diretório corrente.</pre>

<pre>Portanto, execute a célula abaixo para que você tenha a função à sua disposição.</pre>

````python
import requests
import os

# desabilitar a verificação de certificado de segurança para fins didáticos
# é possível, considerando que estamos tratando apenas de dados públicos

requests.packages.urllib3.disable_warnings(requests.packages.urllib3.exceptions.InsecureRequestWarning)

def download_file(url):
    response = requests.get(url, verify = False) 
    
    if response.status_code == 200:
        content = response.content

        if not os.path.exists('data'): os.makedirs('data')

        with open(f'data/{os.path.basename(url)}', mode='wb') as csvfile:
            csvfile.write(content)
````
