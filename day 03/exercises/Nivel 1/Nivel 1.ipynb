### Raspagem Avançada de Dados ✂️

#### BeautifulSoup

<pre>Vamos criar um objeto <b>BeautifulSoup</b> a partir do conteúdo HTML armazenado em <b>response.content</b>, e armazenar na variável <b>soup</b>.</pre>

<details>
  <summary>Resposta</summary>

<br/>

```python
# usamos o conteúdo (response.content) e informamos o formato HTML do conteúdo ('html.parser')
soup = BeautifulSoup(response.content, 'html.parser')
```

<br/>

</details>

<pre>Agora, vamos investigar as tags para encontrar os títulos dos serviços.</pre>

<pre>Ao usar a inspeção do seu navegador, você verá que os serviços mais acessados estarão encapsulados pela tag <b>div</b> com uma das classes sendo <b>servicos-mais-acessados</b>.</pre>

<img src="figs/gov-items.png" width="800px" style="margin-left: auto; margin-right: auto;">

<pre>Usando a variável <b>soup</b>, busque uma tag <b>div</b>, especificando a classe <b>servicos-mais-acessados</b>, e atribua o resultado à variável <b>servicos_mais_acessados</b>.</pre>

👉 dica: use o método `find` da variável `soup` para realizar a busca.

<details>
  <summary>Resposta</summary>

<br/>

```python
servicos_mais_acessados = soup.find('div', class_='servicos-mais-acessados')
```

<br/>

</details>

<pre>Ao inspecionar a variável <b>servicos_mais_acessados</b>, é possível verificar que cada serviço aparece encapsulado por uma tag <b>a</b>, que corresponde a tag de hyperlink do HTML.</pre>

<pre>Vamos, portanto, buscar todas as tags <b>a</b> da variável <b>servicos_mais_acessados</b> e atribuir o resultado à mesma variável <b>servicos_mais_acessados</b>.</pre>

👉 dica: use o método `find_all` da variável `servicos_mais_acessados` para realizar a busca.

<details>
  <summary>Resposta</summary>

<br/>

```python
servicos_mais_acessados = servicos_mais_acessados.find_all('a')
```

<br/>

</details>

<pre>Perceba que cada elemento da lista <b>servicos_mais_acessados</b>, digamos <b>servico_mais_acessado</b>, contém os dados de um dos serviços mais acessados, da seguinte forma:
 - na tag <b>span</b> de classe <b>ranking</b>, temos a posição do <b>ranking</b> que o serviço ocupa;
 - na tag <b>span</b> de classe <b>texto</b>, temos o <b>título</b> dado ao serviço; 
 - no atributo 'href' (<b>servico_mais_acessado['href']</b>), temos o <b>hyperlink</b> para o serviço.</pre>

<img src="figs/gov-items-tags.png" width="800px" style="margin-left: auto; margin-right: auto;">

<pre>Faça, portanto, um percurso na lista <b>servicos_mais_acessados</b> e crie uma lista de dicionários que contenha os campos <b>ranking</b>, <b>titulo</b> e <b>hyperlink</b> para cada serviço.</pre>

👉 dica: use o método `strip` no atributo `text` de cada `find` usado na busca dos dados de cada serviço mais acessado.

<details>
  <summary>Resposta</summary>

<br/>

```python
servicos = []

for servico_mais_acessado in servicos_mais_acessados:
    ranking = servico_mais_acessado.find('span', class_='ranking').text.strip()
    titulo = servico_mais_acessado.find('span', class_='texto').text.strip()
    hyperlink = servico_mais_acessado['href']
    servicos.append({'ranking': ranking, 'titulo': titulo, 'hyperlink': hyperlink})
```
