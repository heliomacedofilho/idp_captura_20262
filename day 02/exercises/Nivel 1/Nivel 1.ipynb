<pre>No SGS, as informações podem ser extraídas por meio de importação direta no site ou podem ser extraídas de forma automatizada.</pre>

### Aspectos de API 📝

<pre>Inicialmente, vamos utilizar a API do SGS para o <a href="https://api.bcb.gov.br/dados/serie/bcdata.sgs.433/dados?formato=json">Índice Nacional de Preços ao Consumidor Amplo</a>.</pre>

<pre>Carregue a série histórica do IPCA, disponível na plataforma do Banco Central do Brasil, usando a URL <b>https://api.bcb.gov.br/dados/serie/bcdata.sgs.433/dados?formato=json</b>, e atribua o resultado da consulta à variável <b>indicador_serie_temporal</b>.</pre>

👉 dica: utilize o método `get` da biblioteca `requests`, passando a URL acima como parâmetro, bem como utilize o método `json` no resultado da operação.

👉 mais dica: na URL, tudo que vem depois do ponto de interrogação (?) são parâmetros da consulta. No caso em tela, temos o par associado `formato` e `json`.
 - Remova, da URL, todos os caracteres a partir do ponto de interrogação (?); e
 - Adicione o dicionário `{"formato": "json"}` ao parâmetro `params` do método `get`.

```python
def obter_indicador_por_data(data):
    url = "https://api.bcb.gov.br/dados/serie/bcdata.sgs.433/dados"
    indicador_serie_temporal = requests.get(url, params={"formato": "json"}).json() 
    for item in indicador_serie_temporal:
        if data in item['data']:
            return item['valor']
    raise Exception("data não encontrada ou fora do padrão")
```

<pre>A consulta à URL <b>https://api.bcb.gov.br/dados/serie/bcdata.sgs.433/dados?formato=json</b> pode ser ainda mais customizada, permitindo que a nossa consulta restrinja o resultado à data desejada.</pre>

<pre>Há algumas razões para customizar a consulta, 
     - uma delas é a economicidade de dados sendo trafegados -- afinal não carece de transmitir toda a série temporal, mas somente o IPCA de uma data --, bem como
     - a delegação do processamento de dados à API, que disponibiliza o serviço, potencialmente usando mecanismos avançados de processamento de dados.</pre>

<pre>Para tanto, no mesmo dicionário que você passou o par <b>"formato": "json"</b>, também passe <b>"dataInicial": data</b> e <b>"dataFinal": data</b>, em que <b>data</b> é o nome do parâmetro que armazena a cadeia de caracteres no formato <b>01/MM/YYYY</b>.</pre>

👉 dica: caso a data não tenha sido encontrada pela API, perceba que a API retornará um dicionário com a chave `erro`. Caso seja encontrada a data, a API retornará uma lista com um elemento/dicionário com o valor do IPCA no campo `valor`.

<details>
  <summary>Resposta</summary>

<br/>

```python
def obter_indicador_por_data(data):
    url = "https://api.bcb.gov.br/dados/serie/bcdata.sgs.433/dados"
    indicador_serie_temporal = requests.get(url, params={"formato": "json", "dataInicial": data, "dataFinal": data}).json()
    if "erro" in indicador_serie_temporal:
        raise Exception("data não encontrada ou fora do padrão")
    else:
        print(indicador_serie_temporal[0]['valor'])
```
