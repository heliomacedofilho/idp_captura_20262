## HTML

<pre>Inicialmente, vamos criar um protótipo de página HTML para entender como funciona sua estrutura.</pre>

```html
<!DOCTYPE html>
<html>
<head>
    <title>Exemplo de Pagina HTML</title>
</head>
<body>
    <h1>Minha Primeira Pagina HTML</h1>
    
    <ul>
        <li><a href="https://www.enap.gov.br/pt/">Home</a></li>
        <li><a href="#">Sobre</a></li>
        <li><a href="#">Contato</a></li>
    </ul>
    
    <p>Bem-vindo 'a minha primeira pagina HTML!</p>
    <p>Esta é uma pagina de exemplo para demonstrar a estrutura basica do HTML.</p>
</body>
</html>
````

<pre>Agora, vamos explicar algumas tags</pre>

```<!DOCTYPE html>```: define o tipo de documento como HTML5.

```<html>...</html>```: define o início e o fim do documento HTML.

```<head>...</head>```: contém informações sobre o documento, como o título da página.

```<title>...</title>```: define o título da página exibido na aba do navegador.

```<body>...</body>```: contém o conteúdo visível da página.

```<h1>...</h1>```: cria um cabeçalho de nível 1 (o maior).

```<ul>...</ul>```: cria uma lista não ordenada.

```<li>...</li>```: define um item da lista.

```<a href="#">...</a>```: cria um link vazio para fins de exemplo.

```<p>...</p>```: cria parágrafos de texto.

🚨 Vamos usar a API <a href='https://dadosabertos.camara.leg.br/swagger/api.html'>Deputados</a>: listagem e busca de deputados, segundo critérios

`````python
response = requests.get('https://dadosabertos.camara.leg.br/'+
                        'api/v2/deputados')
response
`````

<pre>Vamos abrir o conteúdo da resposta.</pre>

`````python
content = response.json() #em vez de response.content
deputados = content['dados']
deputados[:2]
`````

<pre>Alternativamente, vamos averiguar a quantidade de projetos que tramitaram na última semana.</pre>

🚨 Agora, vamos usar a API <a href='https://dadosabertos.camara.leg.br/swagger/api.html'>Proposições</a>: Lista configurável de proposições na Câmara

<pre>Dessa vez, vamos qualificar a resposta com a passagem de parâmetros por meio da API</pre>

<pre>Primeiro, calculamos as datas que passaremos como parâmetros.</pre>

`````python
from datetime import date
from dateutil.relativedelta import relativedelta

hoje = date.today()
semana_passada = hoje + relativedelta(days=-7)
`````

<pre>Agora, consultamos a API, passando os parâmetros, conforme <a href='https://dadosabertos.camara.leg.br/swagger/api.html'>documentação da API</a></pre>

`````python
response = requests.get('https://dadosabertos.camara.leg.br/'+
                        'api/v2/proposicoes', 
                         params={'dataInicio':f'{semana_passada}', 
                                 'dataFim':f'{hoje}', 
                                 'pagina': 1})
content = response.json()
content['dados'][:2]
`````
